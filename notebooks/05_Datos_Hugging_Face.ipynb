## 1. Montar Drive

```python
from google.colab import drive, userdata
drive.mount('/content/drive')

RUTA_DATA = '/content/drive/MyDrive/RUTA/A/data_prep'

print("Drive montado.")
```

## 2. Conexión con Hugging Face

```python
!pip install -U huggingface_hub -q

from huggingface_hub import login, create_repo, upload_file, list_repo_files

login(token=userdata.get('HF_TOKEN'))

TU_USUARIO_HF = "santiagortegasegura"

REPO_DATOS = f"{TU_USUARIO_HF}/emotion-futbol-data"
REPO_CHECKPOINTS = f"{TU_USUARIO_HF}/emotion-futbol-checkpoints"

create_repo(REPO_DATOS, repo_type="dataset", private=True, exist_ok=True)
create_repo(REPO_CHECKPOINTS, repo_type="dataset", private=True, exist_ok=True)

print(f"Listo. Bodega de datos: https://huggingface.co/datasets/{REPO_DATOS}")
print(f"Listo. Bodega de checkpoints: https://huggingface.co/datasets/{REPO_CHECKPOINTS}")
```

## 3. Subir los 3 archivos de datos

```python
ARCHIVOS_A_SUBIR = [
    "04b_entrenamiento_72k.csv",
    "04b_validacion_interna_18k.csv",
    "04b_validacion_final_combinada.csv",
]

for nombre in ARCHIVOS_A_SUBIR:
    ruta_local = f"{RUTA_DATA}/{nombre}"
    upload_file(
        path_or_fileobj=ruta_local,
        path_in_repo=nombre,
        repo_id=REPO_DATOS,
        repo_type="dataset",
    )
    print(f"Subido: {nombre}")

print("\nTodos los archivos subidos.")
```

## 4. Verificar los archivos subidos

```python
archivos_en_hf = list_repo_files(REPO_DATOS, repo_type="dataset")
print("Archivos en tu bodega de datos de Hugging Face:")
for a in archivos_en_hf:
    print(" -", a)
```
